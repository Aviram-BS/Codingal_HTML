```python
print("===========================")
print("Parking LOt")
print("===========================")
def cal_change(paid,price):
         change = paid-price
         return change
parking_price = 50
print("Total Parking lot price :",parking_price)
print("Money notes Accepted: 10,20,30,40")
total_money = 0

while True:
        money = int(input("Insert Money note 10,20,30,40 : "))
        if money!=10 and money!=20 and money!=30 and money!=40 :
            print("Invailid Money")
            continue
        total_money += money
        print("Total money inserted so far",total_money)
        if total_money >= parking_price :
             print("Enough Money is inserted")
             break
change = cal_change(total_money,parking_price)
if change == 0:
     pass
else:
  print("Change due is :",change)

```
